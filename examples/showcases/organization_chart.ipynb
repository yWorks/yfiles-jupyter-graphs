````python
import json
import urllib.request
from yfiles_jupyter_graphs import GraphWidget
from yfiles_jupyter_graphs import NodeStyle, NodeShape, LabelStyle, TextWrapping, TextAlignment, FontWeight
````

````python
url = 'https://raw.githubusercontent.com/yWorks/yfiles-jupyter-graphs/refs/heads/main/examples/showcases/resources/organization_chart_data.json'
with urllib.request.urlopen(url) as f:
    data = json.load(f)

dept_map = {d['id']: d for d in data['departments']}
person_map = {p['id']: p for p in data['people']}
````

````python
# One distinct color per department
DEPT_COLORS = {
    'd-01': '#E63946',  # Executive    - red
    'd-02': '#457B9D',  # Engineering  - steel blue
    'd-03': '#2A9D8F',  # Finance      - teal
    'd-04': '#E9C46A',  # Marketing    - amber
    'd-05': '#F4A261',  # Operations   - orange
    'd-06': '#6A4C93',  # Human Res.   - purple
}

nodes = [
    {
        'id': p['id'],
        'properties': {
            'name': p['name'],
            'role': p['role'],
            'department': dept_map[p['departmentId']]['name'],
            'dept_color': DEPT_COLORS.get(p['departmentId'], '#888888'),
            'text_color': '#1a1a1a',
            'email': p['email'],
            'phone': p['phone'],
        },
    }
    for p in data['people']
]

edges = [
    {
        'start': e['source'],
        'end': e['target'],
    }
    for e in data['edges']
]
````

````python
w = GraphWidget()
w.nodes = nodes
w.edges = edges

# Name + role, text color chosen for contrast against the node background
w.node_label_mapping = lambda n: LabelStyle(
    text=f"{n['properties']['name']}\n{n['properties']['role']}",
    font_size=13,
    font_weight=FontWeight.BOLD,
    color=n['properties']['text_color'],
    wrapping=TextWrapping.WORD,
    text_alignment=TextAlignment.CENTER,
    maximum_width=180,
)

w.node_styles_mapping = lambda n: NodeStyle(
    color=n['properties']['dept_color'],
    shape=NodeShape.ROUND_RECTANGLE,
)
w.node_size_mapping = lambda n: (200, 60)

w.hierarchical_layout()
display(w)
````
