# yFiles Graphs for Jupyter v2.0.0

# Changelog

This version features a greatly improved developer experience with an improved Python API that introduces new convenient data-classes and enums, improved property declarations, and new keyword arguments for the widget constructor.

## Improved AI compatibility
The new [agent-guide.md](https://github.com/yWorks/yfiles-jupyter-graphs/blob/main/agent-guide.md) includes concise information for agents to work with yFiles Graphs for Jupyter. A simple start prompt that worked well in our tests is:

> Read the instructions at https://raw.githubusercontent.com/yWorks/yfiles-jupyter-graphs/refs/heads/main/agent-guide.md. 
> Then, create a new Jupyter notebook that step-by-step explores the <data-file-path> and visualizes it using yfiles-jupyter-graphs. 
> Be creative and design visualizations that highlight interesting insights and aspects of the data.

With the help of this prompt we created the new showcase notebooks (see below).

## Incompatible changes
* `set_<mapping>`, `get_<mapping>`, and `del_<mapping>` function are deprecated now. Instead use the related properties. While these functions still work for backwards-compatibility, they'll emit a warning to switch to the new properties.
* Removing a mapping (formerly `del_<mapping>`) has been replaced by setting the mapping to `None`.
* The widget does not modify the passed user data objects anymore.
* Renamed `hierarchic` layout to `hierarchical`.
* Renamed `hierarchic_layout()` to `hierarchical_layout()`.
* Renamed node shape `round-rectangle` to `round_rectangle`.
* Renamed node shape `hexagon2` to `hexagon_standing`.
* Reworked `TextWrapping` to distinguish between "trim" and "wrap" at maximum size and added the new `TextWrapping.CLIP` option. So the previous text-wrapping options `character`, `character_ellipsis`, `word`, and `word_ellipsis` are replaced by `<trim|wrap>_character`, `<trim|wrap>_character_ellipsis`, `<trim|wrap>_word`, and `<trim|wrap>_word_ellipsis`.
* `widget.selection` now returns a tuple instead of a nested list.
* Minimum required Python version is now 3.7.

## New features
* All properties can now be set conveniently with the `kwargs` of the constructor.
* Added new data-classes wherever the API previously required dictionaries with *correct* property names. However, correctly formatted dictionaries still work as before. The following data-classes are available now:
  * `Node` and `Edge`: Used for structure input data for the widget.
  * `NodeStyle`: May be returned in the `node_styles_mapping`.
  * `EdgeStyle`: May be returned in the `edge_styles_mapping`.
  * `LabelStyle`: May be returned in the `node_label_mapping` or `edge_label_mapping`.
* Added new enums for the following values `Layout`, `NodeShape`, `DashStyle`, `LabelPosition`, `TextWrapping`, `FontWeight`, and `TextAlignment`. Note that the previous string values still work, unless changed by an incompatible change.
* Added new node shape `squircle`, a shape similar to a `pill`, but the shorter sides are rounded with semi-squircles instead of semicircles.
* Modernized the widget's frontend.

## Improvements
* Improved heat map visualization.
* Improved automatic layout when collapsing/expanding group nodes for hierarchical and organic layouts.
* Group node backgrounds are now transparent to potentially show the heatmap of child nodes.
* Updated example notebooks to use the new properties and data-classes.
* Improved and updated the documentation and README.

## Bugfixes
* Fixed pre-selecting nodes with the neighborhood config.
* Fixed nested group nodes becoming larger when a nested group is collapsed/expanded.
* Fixed unnecessary horizontal scrollbar in the sidebar of small widgets.
* Fixed exception with interactive organic layout when starting to drag a label instead of a node.
* Fixed `font` not being applied in label style mappings.
* Fixed unability to return to another graph layout after switching to map layout when not geo-spatial mapping was given.
